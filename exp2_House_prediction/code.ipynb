```python
import matplotlib.pyplot as plt 

import seaborn as sns 

import plotly.express as px
```

```python
plt.rcParams['font.size'] = 14
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
plt.figure(figsize=(8,5))  # Set figure size
plt.plot(df["area"],df["prices"],marker='o',linestyle="-", color="blue")
plt.xlabel("area")   # X-axis label= means the trend is plotted over the order of rows in the dataset.
plt.ylabel("price")  # Y-axis label
plt.title("House_price vs area")  # Title of the plot
plt.show()  # Display the plot
```

```python
plt.scatter(df["area"],df["prices"],marker="+",color="red")
plt.xlabel("area")   # X-axis label= means the trend is plotted over the order of rows in the dataset.
plt.ylabel("price")  # Y-axis label
plt.title("House_price vs area")  # Title of the plot
plt.show()  # Display the plot
```

```python
sns.pairplot(df)  
plt.show()
```

```python
# univariate analysis - visualize distribution of numeric columns
numeric_cols = ['prices', 'area']
for col in numeric_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], kde=True, bins=30, color='blue', alpha=0.6) #alpha = transparency, kde for curve
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
print(df.columns) 
```

```python
df["prices"] = df["prices"].str.replace(",", "").astype(float)
```

```python
model.fit(df[['area']],df['prices'])
```

```python
predicted_price = model.predict([[4500]])
print(predicted_price)
```
